# 04 · Analysis

- **RQ1** Prevalence and completeness of disclosure by regime and platform: pairwise Fisher's exact tests (Holm-corrected), a Monte Carlo exact test for the full table, and Kruskal-Wallis on the index.
- **RQ2** Findability and presentation: Kruskal-Wallis on clicks to odds; presentation and dark-pattern candidate tables.
- **RQ3** Same title across markets: Cochran's Q and McNemar tests, Friedman and Wilcoxon tests on the index, and a logistic regression with standard errors clustered by title.

Unknown codes are dropped from each test, and the number dropped is reported.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
sys.path.insert(0, str(ROOT))
```

```python
import pandas as pd
from analysis import stats
from scrapers.records import PROCESSED

df = pd.read_csv(PROCESSED / "coded_units.csv", dtype={"store_id": str})
```

## RQ1 · Prevalence by regime and platform

```python
r1 = stats.rq1(df)
display(r1["by_regime"], r1["by_platform"])
print("Exact test, regime:", r1["exact_regime"])
print("Exact test, platform:", r1["exact_platform"])
display(r1["pairwise_regime"], r1["pairwise_platform"])
print("PDI by regime:", r1["pdi_kruskal_regime"])
print("PDI by platform:", r1["pdi_kruskal_platform"])
```

## RQ2 · Findability and presentation

```python
r2 = stats.rq2(df)
print(r2["clicks_kruskal_regime"])
display(r2["clicks_summary"])
for name, table in r2["presentation_tables"].items():
    print(name); display(table)
```

## RQ3 · Same title across markets

```python
r3 = stats.rq3(df)
print("Titles in all markets (binary):", r3["n_titles_binary"], "| (PDI):", r3["n_titles_pdi"])
for key in ("cochrans_q", "friedman_pdi"):
    if key in r3: print(key, r3[key])
for key in ("mcnemar_pairwise", "wilcoxon_pairwise"):
    if key in r3: display(r3[key])
logit = r3["logit"]
print(logit if isinstance(logit, str) else logit.summary())
```
